# Urban Green Cover — Exploratory Analysis

Idea 15: Inspect class balance, sample images, and preprocessing effects before training.

```python
import json
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

summary_path = Path("../data/processed/dataset_summary.json")
manifest_path = Path("../data/processed/manifest.csv")

if summary_path.exists():
    summary = json.loads(summary_path.read_text())
    print("Class balance:")
    display(pd.Series(summary["class_balance"]).sort_values(ascending=False))
else:
    print("Run data preparation first: python run_pipeline.py --step prepare")
```

```python
if manifest_path.exists():
    df = pd.read_csv(manifest_path)
    plt.figure(figsize=(8, 4))
    sns.countplot(data=df, x="greenery_class", hue="split", order=sorted(df["greenery_class"].unique()))
    plt.title("Class distribution by split")
    plt.xticks(rotation=20)
    plt.tight_layout()
    plt.show()
```

```python
comparison_path = Path("../outputs/reports/model_comparison.json")
if comparison_path.exists():
    comparison = json.loads(comparison_path.read_text())
    display(pd.DataFrame(comparison).T)
else:
    print("Train and evaluate models first: python run_pipeline.py --step all")
```
